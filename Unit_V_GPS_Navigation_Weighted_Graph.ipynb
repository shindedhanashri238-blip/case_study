# GPS and Navigation System Using a Weighted Graph

## Data Structures Case Study

**Data Structure Used:** Weighted Graph  
**Algorithm Used:** Dijkstra's Shortest Path Algorithm

## 1. Introduction

A navigation system helps users find routes between places. A city can be represented as a graph: locations are vertices and roads are edges. Each road can have a weight such as distance or travel cost.

This case study models a small city map as a weighted graph and uses Dijkstra's algorithm to find the shortest-distance route between two locations.

## 2. Problem Statement

Represent a small city map using a weighted graph. Store the distance between connected locations and find the shortest route from a starting location to a destination using Dijkstra's algorithm.

## 3. Objectives

- Understand graph representation.
- Represent locations as vertices and roads as weighted edges.
- Use an adjacency list to store the graph.
- Implement Dijkstra's shortest path algorithm.
- Display the shortest route and total distance.
- Analyze time and space complexity.

## 4. Theory

**Graph:** A structure consisting of vertices and edges.

**Weighted Graph:** A graph where each edge has a weight, such as distance in kilometres.

**Adjacency List:** Stores each vertex with a list of its neighbouring vertices and edge weights.

**Dijkstra's Algorithm:** Finds shortest distances from a starting vertex in a graph with non-negative edge weights. It repeatedly selects the currently closest unvisited location and updates distances to its neighbours.

## 5. Algorithm

1. Represent each location and road using an adjacency list.
2. Set the start location's distance to 0 and all other distances to infinity.
3. Use a priority queue to select the location with the smallest known distance.
4. For each neighbour, calculate the new possible distance.
5. If the new distance is shorter, update the distance and predecessor.
6. Repeat until the destination is reached or no reachable locations remain.
7. Reconstruct and display the route and its total distance.

```python
import heapq

city_map = {
    "Home": [("School", 4), ("Market", 2)],
    "School": [("Home", 4), ("Hospital", 5), ("Park", 10)],
    "Market": [("Home", 2), ("Park", 3), ("Hospital", 8)],
    "Park": [("Market", 3), ("School", 10), ("Hospital", 2)],
    "Hospital": [("School", 5), ("Market", 8), ("Park", 2)]
}

def dijkstra(graph, start, destination):
    distances = {place: float("inf") for place in graph}
    previous = {place: None for place in graph}
    distances[start] = 0
    priority_queue = [(0, start)]

    while priority_queue:
        current_distance, current_place = heapq.heappop(priority_queue)

        if current_distance != distances[current_place]:
            continue

        if current_place == destination:
            break

        for neighbour, road_distance in graph[current_place]:
            new_distance = current_distance + road_distance
            if new_distance < distances[neighbour]:
                distances[neighbour] = new_distance
                previous[neighbour] = current_place
                heapq.heappush(priority_queue, (new_distance, neighbour))

    if distances[destination] == float("inf"):
        return [], float("inf")

    route = []
    current = destination
    while current is not None:
        route.append(current)
        current = previous[current]
    route.reverse()
    return route, distances[destination]
```

## 6. Demonstration

```python
start_location = "Home"
destination = "Hospital"

route, total_distance = dijkstra(city_map, start_location, destination)

if route:
    print("Starting location:", start_location)
    print("Destination:", destination)
    print("Shortest route:", " -> ".join(route))
    print("Total distance:", total_distance, "km")
else:
    print("No route found.")
```

## 7. Time and Space Complexity

Using an adjacency list and a priority queue, Dijkstra's algorithm typically runs in O((V + E) log V) time, where V is the number of locations and E is the number of roads. The graph and supporting structures require O(V + E) space.

## 8. Advantages

- Represents real-world locations and road distances.
- Finds a shortest-distance route when road weights are non-negative.
- An adjacency list avoids storing entries for every possible pair of locations.
- The graph can be updated to represent additional roads.
