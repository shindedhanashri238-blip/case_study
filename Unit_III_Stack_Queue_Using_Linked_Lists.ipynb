# Implementation of Stack and Queue Operations Using Linked Lists

## Data Structures Case Study

**Data Structures Used:** Stack, Queue and Linked List

## 1. Introduction

Stacks and queues are linear data structures used to organize and process data. A stack follows the Last In, First Out (LIFO) principle, while a queue follows the First In, First Out (FIFO) principle.

A linked list stores data in nodes. Each node contains a value and a link to another node. In this case study, linked lists are used to implement stack and queue operations dynamically.

## 2. Problem Statement

The objective is to implement the basic operations of a stack and a queue using linked lists. The stack should support push, pop and display operations. The queue should support enqueue, dequeue and display operations.

## 3. Objectives

- Understand stack and queue data structures.
- Understand nodes and links in a linked list.
- Implement stack operations using a linked list.
- Implement queue operations using a linked list.
- Observe the output of insertion and deletion operations.
- Analyze the time and space complexity.

## 4. Theory

**Linked List:** A collection of nodes where each node stores data and a reference to the next node.

**Stack:** Follows LIFO. The last inserted item is removed first. `push()` inserts an item and `pop()` removes the top item.

**Queue:** Follows FIFO. The first inserted item is removed first. `enqueue()` inserts at the rear and `dequeue()` removes from the front.

## 5. Algorithm

### Stack
1. Start with an empty top pointer.
2. For push, create a node and make it the new top.
3. For pop, remove the top node and move top to the next node.
4. Display nodes from top to bottom.

### Queue
1. Start with empty front and rear pointers.
2. For enqueue, add a node at the rear.
3. For dequeue, remove the node at the front.
4. If the queue becomes empty, set rear to `None`.
5. Display nodes from front to rear.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class Stack:
    def __init__(self):
        self.top = None

    def push(self, data):
        new_node = Node(data)
        new_node.next = self.top
        self.top = new_node
        print(data, "pushed onto stack")

    def pop(self):
        if self.top is None:
            print("Stack is empty")
            return
        removed = self.top.data
        self.top = self.top.next
        print(removed, "popped from stack")

    def display(self):
        current = self.top
        print("Stack (top to bottom):", end=" ")
        while current:
            print(current.data, end=" ")
            current = current.next
        print()


class Queue:
    def __init__(self):
        self.front = None
        self.rear = None

    def enqueue(self, data):
        new_node = Node(data)
        if self.rear is None:
            self.front = self.rear = new_node
        else:
            self.rear.next = new_node
            self.rear = new_node
        print(data, "added to queue")

    def dequeue(self):
        if self.front is None:
            print("Queue is empty")
            return
        removed = self.front.data
        self.front = self.front.next
        if self.front is None:
            self.rear = None
        print(removed, "removed from queue")

    def display(self):
        current = self.front
        print("Queue (front to rear):", end=" ")
        while current:
            print(current.data, end=" ")
            current = current.next
        print()
```

## 6. Demonstration

```python
# Stack demonstration
stack = Stack()
stack.push(10)
stack.push(20)
stack.push(30)
stack.display()
stack.pop()
stack.display()

print()

# Queue demonstration
queue = Queue()
queue.enqueue(10)
queue.enqueue(20)
queue.enqueue(30)
queue.display()
queue.dequeue()
queue.display()
```

## 7. Time and Space Complexity

For a linked-list implementation:
- Stack `push`: O(1)
- Stack `pop`: O(1)
- Queue `enqueue`: O(1)
- Queue `dequeue`: O(1)
- Display: O(n), where n is the number of nodes
- Space complexity: O(n) to store n elements

## 8. Advantages

- Size can grow or shrink dynamically.
- Push, pop, enqueue and dequeue take O(1) time.
- No fixed capacity is required in advance.
- Memory is allocated as nodes are created.

## 9. Real-World Applications

- Stack: undo operations, browser history and function calls.
- Queue: printer jobs, customer service lines and task scheduling.
- Linked lists: dynamic collections where insertions and deletions are frequent.
