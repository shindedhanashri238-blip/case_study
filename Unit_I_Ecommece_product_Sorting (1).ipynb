## 4. Divide and Conquer Strategy

Divide and Conquer is an algorithmic strategy in which a large problem is
divided into smaller subproblems. Each subproblem is solved separately,
and the results are combined to obtain the final solution.

The three main steps are:

1. **Divide:** Divide the list of products into two smaller parts.
2. **Conquer:** Recursively sort each smaller part.
3. **Combine:** Merge the sorted parts to obtain the final sorted list.

In this case study, Merge Sort is used to apply the Divide and Conquer strategy.

## 5. Merge Sort Algorithm

Merge Sort is a sorting algorithm based on the Divide and Conquer strategy.

### Steps:

1. If the list contains zero or one product, it is already sorted.
2. Find the middle position of the list.
3. Divide the list into two halves.
4. Recursively apply Merge Sort to both halves.
5. Compare the prices of the products in both halves.
6. Merge the products in ascending order of price.
7. Continue until the complete list is sorted.

```python
# E-commerce product data

products = [
    ("Laptop", 55000),
    ("Mobile", 25000),
    ("Headphones", 2000),
    ("Keyboard", 1500),
    ("Monitor", 12000),
    ("Mouse", 800),
    ("Tablet", 18000),
    ("Watch", 5000)
]

print("Original Product List:")

for product in products:
    print(product[0], "₹", product[1])
```

```python
def merge_sort(products):
    if len(products) <= 1:
        return products

    mid = len(products) // 2

    left = merge_sort(products[:mid])
    right = merge_sort(products[mid:])

    return merge(left, right)
```

```python
def merge(left, right):
    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):

        if left[i][1] <= right[j][1]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result
```

```python
# Sort the products using Merge Sort

sorted_products = merge_sort(products)

print("Products Sorted by Price (Low to High):")
print()

for product in sorted_products:
    print(product[0], "₹", product[1])
```

## 6. Time and Space Complexity

### Time Complexity

Merge Sort divides the list into two halves repeatedly and then merges the
sorted halves.

- Best Case: O(n log n)
- Average Case: O(n log n)
- Worst Case: O(n log n)

### Space Complexity

The space complexity of Merge Sort is:

**O(n)**

Additional space is required to store temporary lists during the merging process.
