# Dictionary Application Using Hash Tables

## Data Structures Case Study

**Data Structure Used:** Hash Table  
**Collision Resolution Technique:** Linear Probing

## 1. Introduction

A dictionary application stores words and their meanings. A hash table is a data structure that uses a hash function to calculate where a key should be stored. It can support fast insertion, searching and deletion on average.

Sometimes two keys map to the same index. This is called a collision. In this case study, linear probing is used to find the next available slot.

## 2. Problem Statement

Develop a simple dictionary that stores words and meanings using a hash table. The program should allow the user to insert a word, search for a word and display the stored dictionary entries. Linear probing will be used to handle collisions.

## 3. Objectives

- Understand hash tables and hash functions.
- Store words and meanings as key-value pairs.
- Implement insertion and searching.
- Understand collisions.
- Resolve collisions using linear probing.
- Analyze time and space complexity.

## 4. Theory

**Hash Table:** An array-based structure that stores key-value pairs at indices calculated by a hash function.

**Hash Function:** Converts a key into an array index. This example uses the sum of the character codes modulo the table size.

**Collision:** Occurs when two keys map to the same index.

**Linear Probing:** When a slot is occupied, check the next slot, continuing circularly until an empty slot is found.

## 5. Algorithm

1. Create a hash table with a fixed size.
2. Calculate the index of a word using the hash function.
3. If the calculated slot is empty, store the word and its meaning there.
4. If the slot is occupied by another word, check the next slot (linear probing).
5. To search, calculate the same initial index and probe until the word is found or an empty slot is reached.
6. Display all occupied slots.

```python
class DictionaryHashTable:
    def __init__(self, size=10):
        self.size = size
        self.table = [None] * size

    def hash_function(self, word):
        return sum(ord(char) for char in word.lower()) % self.size

    def insert(self, word, meaning):
        index = self.hash_function(word)
        for _ in range(self.size):
            slot = (index) % self.size
            if self.table[slot] is None:
                self.table[slot] = (word.lower(), meaning)
                print("Inserted:", word)
                return
            if self.table[slot][0] == word.lower():
                self.table[slot] = (word.lower(), meaning)
                print("Updated:", word)
                return
            index += 1
        print("Hash table is full. Cannot insert:", word)

    def search(self, word):
        index = self.hash_function(word)
        for _ in range(self.size):
            slot = index % self.size
            item = self.table[slot]
            if item is None:
                return None
            if item[0] == word.lower():
                return item[1]
            index += 1
        return None

    def display(self):
        print("Dictionary Hash Table:")
        for i, item in enumerate(self.table):
            print(i, ":", item if item is not None else "Empty")
```

## 6. Demonstration

```python
dictionary = DictionaryHashTable(10)
dictionary.insert("apple", "A fruit")
dictionary.insert("book", "A set of written pages")
dictionary.insert("graph", "A collection of vertices and edges")
dictionary.insert("stack", "A LIFO data structure")

print()
print("Meaning of 'book':", dictionary.search("book"))
print("Meaning of 'python':", dictionary.search("python"))
print()
dictionary.display()
```

## 7. Time and Space Complexity

Let n be the number of stored words and m be the table size.
- Average insertion: O(1)
- Average search: O(1)
- Worst-case insertion or search: O(n), when many collisions occur
- Space complexity: O(m) for the table
